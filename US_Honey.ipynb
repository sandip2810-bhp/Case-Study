```python
'''
1. which state is having highest honey production throughout all the year? represent using bar graph ?

'''
```

```python
high_prod=df.groupby('state')['production'].sum().sort_values(ascending=False).reset_index()
high_prod
```

```python
plt.figure(figsize=(10,8))
sns.barplot(data=high_prod,x='production',y='state',palette='hsv')
```

```python
# 2.which state is having highest and lowest yield per colony? represent using bargraph
high_yield_col=df.groupby('state')['yield_per_colony'].mean().sort_values(ascending=False).reset_index()
high_yield_col
```

```python
plt.figure(figsize=(10,8))
sns.barplot(data=high_yield_col,x='yield_per_colony',y='state',palette='hsv')
```

```python
# which year witnesses the highest profit in honey production ? use barplot to represent
```

```python
high_profit=df.groupby('year')['value_of_production'].sum().sort_values(ascending=False).reset_index()
high_profit
```

```python
plt.figure(figsize=(10,8))
sns.barplot(data=high_profit,x='year',y='value_of_production',palette='hsv')
plt.xticks(rotation=45)
plt.show()
```
